# Histogram :

```python
# Used to see the distribution of numerical data : 

plt.figure(figsize=(8,5))
plt.hist(df['Selling_Price'],bins=15)
plt.title("Distribution of Selling Price")
plt.xlabel("Selling_Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Present_Price'],bins=15)
plt.title("Distribution of Present Price")
plt.xlabel("Present_Price")
plt.ylabel("Frequency")
plt.show()
```

# Bar Chart :

```python
# Compare counts of categories : 

fuel_count = df['Fuel_Type'].value_counts()
fuel_count.plot(kind='bar')
plt.title('Fuel Type Count')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Cars')
plt.grid()
plt.show()
```

# Line Chart :

```python
# Show Trends over time :

year_price = df.groupby('Year_C ')['Selling_Price'].mean()
plt.figure(figsize=(10,5))
plt.plot(year_price.index,year_price.values)
plt.title("Average Selling Price by Year")
plt.xlabel("Year_C ")
plt.ylabel("Average Selling Price")
plt.show()
```

# Scatter plot : 

```python
# Find relationship between two numerical variables : 

plt.figure(figsize=(8,5))
plt.scatter(df['Present_Price'],df['Selling_Price'])
plt.title('Present price vs Selling price')
plt.xlabel('Present price')
plt.ylabel('Selling Price')
plt.show()
```

# Pie Chart : 

```python
# Show Percentage Distribution : 

fuel_count = df['Fuel_Type'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(fuel_count,labels = fuel_count.index,autopct = '%1.1f%%')
plt.title('Fuel Type Distribution')
plt.show()
```

# Seaborn Histogram : 

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Selling_Price'],kde=True)
plt.title('Selling Price Distribution')
plt.show()
```

# Seaborn Bar Chart : 

```python
# It is also called "Count Plot" :

plt.figure(figsize=(6,4))
sns.countplot(x='Fuel_Type',data=df)
plt.title('Fuel Type Count')
plt.show()
```

# Heatmap : 

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include='number').corr(),annot=True)
plt.show()
```
